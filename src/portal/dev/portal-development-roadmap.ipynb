## 🎯 Phase 1: Core Functionality Fixes (PRIORITY)

### **Task 1.1: Update Course Configuration System**
**Files to modify:**
- `src/portal/classes.json` → Replace with dynamic discovery
- `src/portal/scripts/portal-manager.js` → Add course discovery functions

**Implementation Steps:**
1. Create `getCourseList()` function to scan course directories
2. Read each course's `course.json` file for metadata
3. Generate unified course catalog

**Acceptance Criteria:**
- [ ] Course list dynamically generated from filesystem
- [ ] All three courses (hphysics, sphysics, natdis) appear correctly
- [ ] Course metadata pulled from individual JSON files

````python
# Task 1.1 - Implementation Template
# JavaScript function to add to portal-manager.js

js_code_template = '''
/**
 * Dynamically discover and load course configurations
 * @returns {Promise<Array>} Array of course objects
 */
async function discoverCourses() {
    const courseDirectories = ['hphysics', 'sphysics', 'natdis'];
    const courses = [];
    
    for (const courseDir of courseDirectories) {
        try {
            const response = await fetch(`./src/portal/${courseDir}/course.json`);
            if (response.ok) {
                const courseData = await response.json();
                courses.push({
                    id: courseDir,
                    ...courseData,
                    portalUrl: `./src/portal/${courseDir}/portal.html`
                });
            }
        } catch (error) {
            console.warn(`Failed to load course ${courseDir}:`, error);
        }
    }
    
    return courses;
}
'''

print("JavaScript Template for Dynamic Course Discovery:")
print(js_code_template)
````

### **Task 1.2: Implement Dynamic Course Loading**
**Files to modify:**
- `src/portal/index.html` → Update course section JavaScript
- Remove hardcoded course information

**Implementation Steps:**
1. Replace static course cards with dynamic generation
2. Use course JSON data for titles, descriptions, progress
3. Implement proper error handling for missing courses

**Acceptance Criteria:**
- [ ] Course cards generated from JSON data
- [ ] Proper fallbacks for missing course data
- [ ] Consistent styling across all courses

````python
# Task 1.2 - Course Card Generation Template

course_card_template = '''
/**
 * Generate course card HTML from course data
 * @param {Object} course - Course data object
 * @returns {string} HTML string for course card
 */
function generateCourseCard(course) {
    const progress = getProgress(course.id);
    const completedUnits = getCompletedUnits(course.id);
    const totalUnits = course.units ? course.units.length : getTotalUnits(course.id);
    
    return `
        <div class="course-card" data-course="${course.id}">
            <div class="course-header">
                <h3 class="course-title">${course.title || course.curriculumTitle}</h3>
                <div class="course-level">${course.level || 'Standard'}</div>
            </div>
            <div class="course-description">
                ${course.description || course.courseDescription}
            </div>
            <div class="course-progress">
                <div class="progress-bar">
                    <div class="progress-fill" style="width: ${progress}%"></div>
                </div>
                <div class="progress-text">${completedUnits}/${totalUnits} units completed</div>
            </div>
            <div class="course-actions">
                <button onclick="window.location.href='${course.portalUrl}'" class="btn-primary">
                    Enter Course
                </button>
                <button onclick="viewCourseInfo('${course.id}')" class="btn-secondary">
                    Course Info
                </button>
            </div>
        </div>
    `;
}
'''

print("Course Card Generation Template:")
print(course_card_template)
````
