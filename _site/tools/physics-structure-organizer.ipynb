# Physics Curriculum Structure Organizer

This notebook systematically organizes and populates the Cosmos in the Classroom physics curriculum structure, addressing the validation warnings and creating a complete, consistent directory hierarchy.

## Objectives:
1. Clean up duplicate unit directories (01_unit1 vs 01_principles)
2. Create missing structural directories (lessons, labs, problemsets)
3. Generate template README.md files for all units
4. Establish consistent naming conventions
5. Create content templates for future development

```python
import os
import json
import shutil
from pathlib import Path
from datetime import datetime

# Base configuration
BASE_DIR = Path(r'D:\python\Jupyter\build\cosmosintheclassroom')
PHYSICS_DIR = BASE_DIR / 'physics'
HONORS_DIR = PHYSICS_DIR / 'honors' / 'units'
STANDARD_DIR = PHYSICS_DIR / 'standard' / 'units'

print(f"Working directory: {BASE_DIR}")
print(f"Physics directory: {PHYSICS_DIR}")
print(f"Honors units: {HONORS_DIR}")
print(f"Standard units: {STANDARD_DIR}")
```

First, let's establish the correct unit naming and organization based on the Honors Physics curriculum.

## Step 2: Clean Up Existing Directory Structure

Remove the old generic unit directories and organize existing content properly.

```python
def clean_old_directories():
    """Remove old generic unit directories and organize content."""
    
    # List of ALL old directories to remove - be aggressive
    old_dirs = [
        '01_unit1', '02_unit2', '03_unit3', '04_unit4', '05_unit5',
        '06_unit6', '07_unit7', '08_unit8', '09_unit9',
        '02_kinematics1', '03_kinematics2', '05_conservation', 
        '07_shm', '08_electromagnetism', '09_electricity', '10_optics'
    ]
    
    removed_dirs = []
    
    # Clean honors directories
    for old_dir in old_dirs:
        old_path = HONORS_DIR / old_dir
        if old_path.exists():
            print(f"Removing old honors directory: {old_dir}")
            shutil.rmtree(old_path)
            removed_dirs.append(f"honors/{old_dir}")
    
    # Clean standard directories
    for old_dir in old_dirs:
        old_path = STANDARD_DIR / old_dir
        if old_path.exists():
            print(f"Removing old standard directory: {old_dir}")
            shutil.rmtree(old_path)
            removed_dirs.append(f"standard/{old_dir}")
    
    return removed_dirs

# Clean up old directories
removed = clean_old_directories()
print(f"\nRemoved {len(removed)} old directories")

# Check current state
print("\n📁 CURRENT DIRECTORY STATE:")
if HONORS_DIR.exists():
    honors_units = sorted([d.name for d in HONORS_DIR.iterdir() if d.is_dir()])
    print(f"Honors units ({len(honors_units)}): {honors_units}")
else:
    print("Honors directory doesn't exist yet")

if STANDARD_DIR.exists():
    standard_units = sorted([d.name for d in STANDARD_DIR.iterdir() if d.is_dir()])
    print(f"Standard units ({len(standard_units)}): {standard_units}")
else:
    print("Standard directory doesn't exist yet")
```

## Step 3: Create Complete Directory Structure

Create all missing directories and subdirectories for both honors and standard courses.

```python
def create_unit_structure(course_dir, unit_id, unit_info):
    """Create complete directory structure for a unit."""
    
    unit_path = course_dir / unit_id
    unit_path.mkdir(parents=True, exist_ok=True)
    
    # Create required subdirectories
    subdirs = ['lessons', 'labs', 'problemsets', 'resources']
    
    for subdir in subdirs:
        subdir_path = unit_path / subdir
        subdir_path.mkdir(exist_ok=True)
        
        # Create .gitkeep file to ensure empty directories are tracked
        gitkeep_path = subdir_path / '.gitkeep'
        if not gitkeep_path.exists():
            gitkeep_path.write_text('# This file ensures the directory is tracked by Git\n')
    
    return unit_path

def create_all_units():
    """Create directory structure for all units in both courses."""
    
    created_units = {'honors': [], 'standard': []}
    
    # Create honors units
    for unit_id, unit_info in PHYSICS_UNITS.items():
        unit_path = create_unit_structure(HONORS_DIR, unit_id, unit_info)
        created_units['honors'].append(unit_id)
        print(f"Created honors unit: {unit_id}")
    
    # Create standard units (same structure, different content level)
    for unit_id, unit_info in PHYSICS_UNITS.items():
        unit_path = create_unit_structure(STANDARD_DIR, unit_id, unit_info)
        created_units['standard'].append(unit_id)
        print(f"Created standard unit: {unit_id}")
    
    return created_units

# Create all unit structures
created = create_all_units()
print(f"\nCreated {len(created['honors'])} honors units and {len(created['standard'])} standard units")
```

## Step 4: Generate README.md Templates

Create comprehensive README.md files for each unit with proper structure and content.

```python
def generate_unit_readme(unit_id, unit_info, course_level):
    """Generate a comprehensive README.md for a unit."""
    
    readme_content = f"""# {unit_info['title']}

**Course Level**: {course_level.title()} Physics  
**Unit**: {unit_id}  
**Duration**: 3-4 weeks  
**Last Updated**: {datetime.now().strftime('%B %d, %Y')}

## Overview

{unit_info['description']}

## Learning Objectives

By the end of this unit, students will be able to:

- Understand and apply fundamental concepts related to {unit_info['title'].lower()}
- Solve problems involving the key principles covered in this unit
- Connect theoretical concepts to real-world applications
- Demonstrate understanding through laboratory investigations
- Communicate physics concepts using appropriate scientific vocabulary

## Key Topics

""" + '\n'.join([f"- {topic}" for topic in unit_info['topics']]) + f"""

## Unit Structure

### Lessons (`lessons/`)
- **Lesson 1**: Introduction and fundamental concepts
- **Lesson 2**: Core principles and mathematical relationships
- **Lesson 3**: Applications and problem-solving strategies
- **Lesson 4**: Advanced topics and synthesis

### Laboratory Investigations (`labs/`)
- Hands-on experiments to explore key concepts
- Data collection and analysis activities
- Scientific method and inquiry-based learning

### Problem Sets (`problemsets/`)
- Conceptual understanding problems
- Mathematical application exercises
- Real-world scenario challenges
- Assessment preparation materials

### Resources (`resources/`)
- Reference materials and external links
- Supplementary readings
- Video resources and simulations
- Assessment rubrics and answer keys

## Prerequisites

Students should have completed:
- Previous physics units in sequence
- Appropriate mathematics background (Algebra II/Trigonometry)
- Basic scientific method and laboratory skills

## Assessment Methods

- **Formative Assessment**: Daily problems, lab reports, discussions
- **Summative Assessment**: Unit tests, lab practicals, projects
- **Performance Indicators**: Problem-solving accuracy, lab technique, conceptual understanding

## Connections to Other Units

This unit builds upon concepts from previous units and provides foundation for:
- Subsequent physics topics
- Advanced placement physics courses
- STEM career pathways
- Real-world applications in technology and engineering

## Resources and References

- OpenStax Physics Textbook
- PhET Interactive Simulations
- Khan Academy Physics
- Laboratory equipment and safety guidelines

---

*This unit is part of the Cosmos in the Classroom physics curriculum, designed to provide comprehensive, engaging physics education for high school students.*
"""
    
    return readme_content

def create_all_readmes():
    """Create README.md files for all units in both courses."""
    
    created_readmes = []
    
    # Create honors README files
    for unit_id, unit_info in PHYSICS_UNITS.items():
        readme_path = HONORS_DIR / unit_id / 'README.md'
        readme_content = generate_unit_readme(unit_id, unit_info, 'honors')
        
        readme_path.write_text(readme_content, encoding='utf-8')
        created_readmes.append(f"honors/{unit_id}")
        print(f"Created README for honors/{unit_id}")
    
    # Create standard README files
    for unit_id, unit_info in PHYSICS_UNITS.items():
        readme_path = STANDARD_DIR / unit_id / 'README.md'
        readme_content = generate_unit_readme(unit_id, unit_info, 'standard')
        
        readme_path.write_text(readme_content, encoding='utf-8')
        created_readmes.append(f"standard/{unit_id}")
        print(f"Created README for standard/{unit_id}")
    
    return created_readmes

# Create all README files
readmes = create_all_readmes()
print(f"\nCreated {len(readmes)} README files")
```

## Step 5: Create Content Templates

Generate template files for lessons, labs, and problem sets to guide future content development.

```python
def create_lesson_template(unit_id, lesson_num=1):
    """Create a template lesson file."""
    
    template = f"""---
title: "Lesson {lesson_num}: Introduction to {PHYSICS_UNITS[unit_id]['title']}"
unit: "{unit_id}"
lesson: {lesson_num}
course: "honors" # or "standard"
duration: "50 minutes"
objectives:
  - "Students will understand..."
  - "Students will be able to..."
prerequisites:
  - "Previous lesson concepts"
materials:
  - "Textbook"
  - "Calculator"
  - "Lab equipment (if applicable)"
---

# Lesson {lesson_num}: Introduction to {PHYSICS_UNITS[unit_id]['title']}

## Learning Objectives
- Students will understand the fundamental concepts of {PHYSICS_UNITS[unit_id]['title'].lower()}
- Students will be able to apply these concepts to solve basic problems
- Students will connect these ideas to real-world applications

## Prerequisites
- Understanding of previous unit concepts
- Basic mathematical skills

## Introduction (10 minutes)
### Hook
[Engaging opening activity or demonstration]

### Objectives Review
[Present lesson objectives to students]

## Core Content (25 minutes)
### Key Concepts
1. **Concept 1**: [Explanation]
2. **Concept 2**: [Explanation]
3. **Concept 3**: [Explanation]

### Mathematical Relationships
- **Equation 1**: $equation = here$
- **Equation 2**: $another = equation$

### Examples
#### Example 1: [Problem Type]
**Given**: [Known values]  
**Find**: [Unknown values]  
**Solution**: [Step-by-step solution]

## Practice and Application (10 minutes)
### Guided Practice
[Problems to work through together]

### Independent Practice
[Problems for students to attempt]

## Closure (5 minutes)
### Summary
[Key takeaways from the lesson]

### Preview
[Connection to next lesson]

## Homework/Extensions
- Read textbook sections [X.X - X.X]
- Complete practice problems [list specific problems]
- [Optional enrichment activities]

## Assessment
- **Formative**: [Ongoing checks for understanding]
- **Summative**: [End-of-lesson assessment]

## Differentiation
- **Support**: [Accommodations for struggling students]
- **Extension**: [Challenges for advanced students]

## Resources
- Textbook: Chapter [X], Sections [X.X - X.X]
- Online simulations: [Links]
- Video resources: [Links]
"""
    
    return template

def create_lab_template(unit_id, lab_num=1):
    """Create a template lab file."""
    
    template = f"""---
title: "Lab {lab_num}: {PHYSICS_UNITS[unit_id]['title']} Investigation"
unit: "{unit_id}"
lab: {lab_num}
course: "honors" # or "standard"
duration: "90 minutes" # or "50 minutes"
type: "investigation" # or "verification", "design"
safety_level: "low" # or "medium", "high"
group_size: "2-3 students"
---

# Lab {lab_num}: {PHYSICS_UNITS[unit_id]['title']} Investigation

## Purpose
To investigate [specific phenomenon or relationship] and develop understanding of {PHYSICS_UNITS[unit_id]['title'].lower()} through hands-on experimentation.

## Learning Objectives
- Students will measure and analyze [specific quantities]
- Students will identify relationships between [variables]
- Students will practice proper laboratory techniques and safety procedures
- Students will communicate results through data analysis and scientific writing

## Background
[Brief theoretical background relevant to the lab]

### Key Equations
- $equation_1 = relationship$
- $equation_2 = another$

## Materials
- [List all equipment and materials needed]
- [Include quantities and specifications]
- [Note any safety equipment required]

## Safety Considerations
- [List all safety precautions]
- [Identify potential hazards]
- [Specify protective equipment]

## Procedure
### Setup
1. [Detailed setup instructions]
2. [Include diagrams if helpful]
3. [Safety check procedures]

### Data Collection
1. [Step-by-step measurement procedures]
2. [Specify number of trials]
3. [Data recording guidelines]

### Cleanup
1. [Equipment return procedures]
2. [Safety shutdown steps]
3. [Waste disposal guidelines]

## Data Tables
### Table 1: [Description]
| Trial | Variable 1 | Variable 2 | Calculated Value |
|-------|------------|------------|------------------|
| 1     |            |            |                  |
| 2     |            |            |                  |
| 3     |            |            |                  |

## Analysis Questions
1. **Graph Creation**: Create a graph of [x-variable] vs [y-variable]. What is the relationship?

2. **Calculations**: Calculate [specific quantities] using your data.

3. **Error Analysis**: What are the main sources of uncertainty in your measurements?

4. **Theoretical Comparison**: How do your results compare to theoretical predictions?

5. **Real-World Applications**: How does this investigation relate to [real-world phenomenon]?

## Conclusion
Write a brief conclusion that addresses:
- What you learned about [specific concepts]
- How well your data supported the theoretical relationships
- Suggestions for improving the investigation

## Extensions
- [Optional advanced investigations]
- [Connections to other physics concepts]
- [Research opportunities]

## Rubric
| Criteria | Excellent (4) | Proficient (3) | Developing (2) | Beginning (1) |
|----------|---------------|----------------|----------------|---------------|
| Data Collection | [Description] | [Description] | [Description] | [Description] |
| Analysis | [Description] | [Description] | [Description] | [Description] |
| Conclusion | [Description] | [Description] | [Description] | [Description] |
| Lab Skills | [Description] | [Description] | [Description] | [Description] |
"""
    
    return template

def create_problemset_template(unit_id, ps_num=1):
    """Create a template problem set file."""
    
    template = f"""---
title: "Problem Set {unit_id}.{ps_num}: {PHYSICS_UNITS[unit_id]['title']}"
unit: "{unit_id}"
problemset: "{ps_num}"
course: "honors" # or "standard"
difficulty: "intermediate" # or "beginner", "advanced"
estimated_time: "45 minutes"
topics:
  - "{PHYSICS_UNITS[unit_id]['topics'][0]}"
  - "{PHYSICS_UNITS[unit_id]['topics'][1] if len(PHYSICS_UNITS[unit_id]['topics']) > 1 else 'Applications'}"
---

# Problem Set {unit_id}.{ps_num}: {PHYSICS_UNITS[unit_id]['title']}

**Course**: Honors Physics  
**Unit**: {unit_id}  
**Estimated Time**: 45 minutes  
**Due Date**: [Insert due date]

## Instructions
- Show all work clearly and organize your solutions
- Include units in all calculations and final answers
- Use appropriate significant figures
- Draw diagrams where helpful
- Box or highlight your final answers

## Conceptual Questions

### Problem 1
[Conceptual question about fundamental concepts]

### Problem 2
[Question requiring explanation of physics principles]

## Calculation Problems

### Problem 3
**Given**: [List known quantities]  
**Find**: [List unknown quantities]  

[Problem statement with context]

### Problem 4
**Given**: [List known quantities]  
**Find**: [List unknown quantities]  

[Problem statement with context]

### Problem 5
**Given**: [List known quantities]  
**Find**: [List unknown quantities]  

[Problem statement with context]

## Application Problems

### Problem 6
[Real-world scenario requiring physics analysis]

### Problem 7
[Multi-step problem combining several concepts]

## Challenge Problems (Optional)

### Problem 8
[Advanced problem for enrichment]

### Problem 9
[Extension problem connecting to other units]

## Reflection Questions

1. Which problem was most challenging for you? Why?

2. How do the concepts in this problem set connect to real-world situations?

3. What questions do you still have about {PHYSICS_UNITS[unit_id]['title'].lower()}?

---

**Answer Key**: Available in the resources directory  
**Help Resources**: Textbook Chapter [X], Khan Academy, Office Hours
"""
    
    return template

# Create sample templates for the first unit
sample_unit = '01_principles'

# Create lesson template
lesson_template = create_lesson_template(sample_unit)
lesson_path = HONORS_DIR / sample_unit / 'lessons' / 'lesson01_template.md'
lesson_path.write_text(lesson_template, encoding='utf-8')
print(f"Created lesson template: {lesson_path}")

# Create lab template
lab_template = create_lab_template(sample_unit)
lab_path = HONORS_DIR / sample_unit / 'labs' / 'lab01_template.md'
lab_path.write_text(lab_template, encoding='utf-8')
print(f"Created lab template: {lab_path}")

# Create problem set template
ps_template = create_problemset_template(sample_unit)
ps_path = HONORS_DIR / sample_unit / 'problemsets' / 'ps01.1_template.md'
ps_path.write_text(ps_template, encoding='utf-8')
print(f"Created problem set template: {ps_path}")

print("\nTemplate files created for reference and future development")
```

## Step 6: Migration Status Report

Generate a comprehensive report of what has been created and what still needs attention.

```python
def generate_status_report():
    """Generate a comprehensive status report of the physics curriculum organization."""
    
    report = {
        'timestamp': datetime.now().isoformat(),
        'structure_complete': True,
        'honors_units': {},
        'standard_units': {},
        'missing_content': [],
        'next_steps': []
    }
    
    # Check honors units
    for unit_id in PHYSICS_UNITS.keys():
        unit_path = HONORS_DIR / unit_id
        if unit_path.exists():
            report['honors_units'][unit_id] = {
                'exists': True,
                'readme': (unit_path / 'README.md').exists(),
                'lessons_dir': (unit_path / 'lessons').exists(),
                'labs_dir': (unit_path / 'labs').exists(),
                'problemsets_dir': (unit_path / 'problemsets').exists(),
                'resources_dir': (unit_path / 'resources').exists()
            }
        else:
            report['honors_units'][unit_id] = {'exists': False}
            report['missing_content'].append(f"honors/{unit_id}")
    
    # Check standard units
    for unit_id in PHYSICS_UNITS.keys():
        unit_path = STANDARD_DIR / unit_id
        if unit_path.exists():
            report['standard_units'][unit_id] = {
                'exists': True,
                'readme': (unit_path / 'README.md').exists(),
                'lessons_dir': (unit_path / 'lessons').exists(),
                'labs_dir': (unit_path / 'labs').exists(),
                'problemsets_dir': (unit_path / 'problemsets').exists(),
                'resources_dir': (unit_path / 'resources').exists()
            }
        else:
            report['standard_units'][unit_id] = {'exists': False}
            report['missing_content'].append(f"standard/{unit_id}")
    
    # Identify next steps
    report['next_steps'] = [
        'Migrate existing content from src/hphys and src/sphys directories',
        'Create actual lesson content using the templates provided',
        'Develop laboratory investigations for each unit',
        'Create comprehensive problem sets with answer keys',
        'Update portal navigation to reference new physics structure',
        'Create Marp slide presentations for each unit',
        'Develop assessment materials and rubrics'
    ]
    
    return report

# Generate and display status report
status = generate_status_report()

print("📊 PHYSICS CURRICULUM ORGANIZATION STATUS REPORT")
print("=" * 60)
print(f"Generated: {status['timestamp']}")
print(f"Structure Complete: {'✅' if status['structure_complete'] else '❌'}")

print("\n🎓 HONORS PHYSICS UNITS:")
for unit_id, info in status['honors_units'].items():
    if info['exists']:
        status_icon = '✅' if all([info['readme'], info['lessons_dir'], info['labs_dir'], info['problemsets_dir']]) else '⚠️'
        print(f"  {status_icon} {unit_id}: {PHYSICS_UNITS[unit_id]['title']}")
    else:
        print(f"  ❌ {unit_id}: Missing")

print("\n📚 STANDARD PHYSICS UNITS:")
for unit_id, info in status['standard_units'].items():
    if info['exists']:
        status_icon = '✅' if all([info['readme'], info['lessons_dir'], info['labs_dir'], info['problemsets_dir']]) else '⚠️'
        print(f"  {status_icon} {unit_id}: {PHYSICS_UNITS[unit_id]['title']}")
    else:
        print(f"  ❌ {unit_id}: Missing")

if status['missing_content']:
    print(f"\n⚠️  Missing Content: {len(status['missing_content'])} items")
    for item in status['missing_content']:
        print(f"    - {item}")

print("\n🚀 NEXT STEPS:")
for i, step in enumerate(status['next_steps'], 1):
    print(f"  {i}. {step}")

# Save status report to file
status_file = BASE_DIR / 'physics' / 'shared' / 'archive' / 'organization-status.json'
status_file.parent.mkdir(parents=True, exist_ok=True)
with open(status_file, 'w') as f:
    json.dump(status, f, indent=2)

print(f"\n📋 Status report saved to: {status_file}")
```
